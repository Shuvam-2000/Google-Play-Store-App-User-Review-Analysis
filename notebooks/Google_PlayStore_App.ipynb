The dataset initially contained 483 exact duplicate rows.
Inspection showed that these records were identical across all columns, indicating redundant copies of the same observation.
Keeping these duplicate rows would unnecessarily increase the dataset size and could introduce bias in the analysis.
Therefore, the duplicate rows were removed using drop_duplicates() to ensure that each record appears only once.

```python
# checking for duplicate values after removing the duplicaties
data['App'].duplicated().sum()
```

## Observation 2: Duplicate App Names
After removing exact duplicate rows, duplicate app names were still present in the dataset.
Inspection revealed that these records represented the same application, with almost identical information across all columns.
The primary difference among these entries was the Reviews column, where review counts varied slightly. This suggests that the same app was captured multiple times at different stages, with updated review counts.
To avoid counting the same application multiple times, the dataset was sorted by the Reviews column in descending order, and for each app, the record with the highest number of reviews was retained while the remaining duplicate entries were removed.
This approach preserves the most recent and complete record for each application, resulting in a cleaner and more reliable dataset for further analysis.

```python
data['Size'].unique()
```

```python
# convert the dtype and restructure the values
def convert_size(size):
    if pd.isna(size):
        return np.nan
    elif size.endswith('M'):
        return float(size[:-1])
    elif size.endswith('k'):
        return float(size[:-1]) / 1024
    else:
        return np.nan

data['Size'] = data['Size'].apply(convert_size)
```

```python
# check the dtype after conversion 
data['Size'].dtype
```

```python
data['Size'].head()
```

```python
data['Size'].unique()
```

```python
data.info()
```

## Observation:
The Size column contained values in both megabytes (M) and kilobytes (k), along with the non-numeric value "Varies with device". To ensure consistency, all sizes were converted to megabytes (MB). Entries labeled "Varies with device" were replaced with missing values (NaN) because they do not represent a fixed application size. The cleaned column was then converted to the float data type, making it suitable for numerical analysis.

## Check Outliers

```python
data.info()
```

```python
# checking for outliers for Rating 
# Using the IQR method
Q1 = data['Rating'].quantile(0.25)
Q3 = data['Rating'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
print(lower_bound)
upper_bound = Q3 + 1.5 * IQR

outliers = data[(data['Rating'] < lower_bound) | (data['Rating'] > upper_bound)]

print("Number of outliers:", len(outliers))
```

```python
# plotiing the outliers in rating to visualize the spread of outliers
plt.figure(figsize=(8, 4))

sns.boxplot(x=data['Rating'].dropna(), color='skyblue', linewidth=1.5)

plt.title("Boxplot of Rating", fontsize=14, fontweight='bold')
plt.xlabel("Rating", fontsize=12)

plt.tight_layout()
plt.show()
```

## Observation
Using the IQR method, 491 outliers were identified in the Rating column. All of these observations were below the lower bound of 3.25, while no upper outliers were detected because app ratings are naturally capped at 5.0. These lower values represent genuinely low-rated applications rather than data entry errors. Therefore, the outliers were retained to preserve the true distribution of app ratings and avoid introducing bias into the analysis.
